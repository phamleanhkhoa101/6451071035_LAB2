**Đề bài:** Lập trình tính giá trị F-measure (F1-score) của các bài tập cơ bản.
$$F\text{-measure} = 2 \times \frac{\text{precision} \times \text{recall}}{\text{precision} + \text{recall}}$$
**Họ và tên:** Phạm Lê Anh Khoa  
**MSSV:** 6451071035

### Câu 1: Định nghĩa hàm tính F-measure từ Precision và Recall

```python
print("Phạm Lê Anh Khoa - 6451071035")

def calculate_f_measure(precision, recall):
    """
    Hàm tính giá trị F-measure (F1-score)
    Công thức: F = 2 * (precision * recall) / (precision + recall)
    """
    if precision + recall == 0:
        return 0.0
    return 2 * (precision * recall) / (precision + recall)
```

### Câu 2: Tính F-measure cho Bài tập cơ bản 1: Đánh giá rủi ro hồ sơ vay tín dụng (Trang 10 - 15)
- **Cây quyết định (CART)**: Precision = 2/3 (66.67%), Recall = 2/2 (100%)
- **Naïve Bayes**: Precision = 2/2 (100%), Recall = 2/2 (100%)

```python
print("Phạm Lê Anh Khoa - 6451071035")

# 1. Mô hình Cây quyết định (CART)
p_cart = 2 / 3
r_cart = 2 / 2
f_cart = calculate_f_measure(p_cart, r_cart)

# 2. Mô hình Naïve Bayes
p_nb = 2 / 2
r_nb = 2 / 2
f_nb = calculate_f_measure(p_nb, r_nb)

print("=== KẾT QUẢ BÀI TẬP VAY TÍN DỤNG (TRANG 10 - 15) ===")
print(f"[Cây quyết định CART]  Precision: {p_cart:.4f} | Recall: {r_cart:.4f} => F-measure: {f_cart:.4f} (~{f_cart*100:.2f}%)")
print(f"[Thuật toán Naïve Bayes] Precision: {p_nb:.4f} | Recall: {r_nb:.4f} => F-measure: {f_nb:.4f} (~{f_nb*100:.2f}%)")
```

### Câu 3: Hàm tổng quát đánh giá F-measure từ ma trận nhầm lẫn (Confusion Matrix)
Áp dụng tính toán nhanh cho các bài toán phân loại nhị phân và đa lớp.

```python
print("Phạm Lê Anh Khoa - 6451071035")

import pandas as pd

def evaluate_from_cm(tp, fp, fn, tn, model_name="Model"):
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f_measure = calculate_f_measure(precision, recall)
    return {
        "Mô hình": model_name,
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F-measure": round(f_measure, 4)
    }

# Bảng tổng hợp so sánh F-measure
results = []
results.append(evaluate_from_cm(tp=2, fp=1, fn=0, tn=0, model_name="Vay tín dụng - Cây CART"))
results.append(evaluate_from_cm(tp=2, fp=0, fn=0, tn=1, model_name="Vay tín dụng - Naïve Bayes"))

df_results = pd.DataFrame(results)
display(df_results)
```
