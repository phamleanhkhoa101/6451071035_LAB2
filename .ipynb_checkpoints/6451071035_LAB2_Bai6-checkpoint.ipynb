### Chuẩn bị: Import các thư viện cần thiết và nạp dữ liệu

```python
print("Phạm Lê Anh Khoa - 6451071035")

%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics

# Đọc dữ liệu từ file mushrooms.csv
df = pd.read_csv('Dataset/mushrooms.csv')
print(f"Kích thước dữ liệu: {df.shape}")
df.info()
```

### Câu 4: Tách các cột dữ liệu thành hai phần (`features` và `labels`)
Thuộc tính quyết định là `'class'` (`p`: poisonous - nấm độc, `e`: edible - ăn được).

```python
# Tách thuộc tính đặc trưng và thuộc tính quyết định
features = df.drop('class', axis=1)
labels = df['class']
```

### Câu 5: Chuyển đổi các cột không phải dạng số về dạng one-hot vector

```python
print("Phạm Lê Anh Khoa - 6451071035")

# Kiểm tra các cột định tính (truyền thêm 'str' để tắt cảnh báo Pandas4Warning)
cat_cols = features.select_dtypes(include=['object', 'str']).columns
print(f"Số lượng cột định tính cần mã hóa: {len(cat_cols)}")

# Mã hóa one-hot vector bằng pd.get_dummies
features_onehot = pd.get_dummies(features, columns=cat_cols)
print(f"Kích thước dữ liệu sau khi mã hóa One-hot: {features_onehot.shape}")
features_onehot.head()
```

### Câu 6: Tách các dòng dữ liệu ra thành hai phần huấn luyện (70%) và kiểm thử (30%)

```python
print("Phạm Lê Anh Khoa - 6451071035")

X_train, X_test, y_train, y_test = train_test_split(
    features_onehot, labels, test_size=0.3, random_state=42
)

print(f"Số lượng dữ liệu huấn luyện: {len(X_train)}")
print(f"Số lượng dữ liệu kiểm thử: {len(X_test)}")
```

### Câu 7: Xây dựng cây ID3 dựa trên dữ liệu huấn luyện, kiểm thử bằng ma trận nhầm lẫn và biểu diễn cây

```python
print("Phạm Lê Anh Khoa - 6451071035")

clf_id3 = tree.DecisionTreeClassifier(criterion="entropy", random_state=0)
clf_id3.fit(X_train, y_train)
```

```python
print("Phạm Lê Anh Khoa - 6451071035")

id3_pred = clf_id3.predict(X_test)
id3_score = metrics.accuracy_score(y_test, id3_pred)
print(f"ID3 Accuracy: {id3_score:.4f} (~{id3_score*100:.2f}%)")
print("\nReport:")
print(metrics.classification_report(y_test, id3_pred))
```

```python
print("Phạm Lê Anh Khoa - 6451071035")

id3_cm = metrics.confusion_matrix(y_test, id3_pred)
class_names = ['edible (e)', 'poisonous (p)']

plt.figure(figsize=(7, 6))
sns.heatmap(id3_cm, annot=True, fmt="d", square=True, cmap='Blues_r',
            xticklabels=class_names, yticklabels=class_names)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title(f'Decision Tree (ID3) Accuracy Score: {id3_score:.4f}', size=13)
plt.show()
```

```python
print("Phạm Lê Anh Khoa - 6451071035")

# Cây quyết định phân loại nấm rất tinh gọn, hiển thị toàn bộ cây hoặc giới hạn max_depth=5
fig, ax = plt.subplots(figsize=(24, 10))
tree.plot_tree(
    clf_id3, 
    max_depth=5, 
    feature_names=features_onehot.columns, 
    class_names=['e', 'p'], 
    filled=True, 
    fontsize=9
)
plt.savefig('mushroom_decision_tree_id3.png', dpi=100, bbox_inches='tight')
plt.show()
```

### Câu 8: Lặp lại Câu 7 đối với cây CART (`criterion='gini'`)

```python
print("Phạm Lê Anh Khoa - 6451071035")

clf_cart = tree.DecisionTreeClassifier(criterion="gini", random_state=0)
clf_cart.fit(X_train, y_train)
```

```python
print("Phạm Lê Anh Khoa - 6451071035")

cart_pred = clf_cart.predict(X_test)
cart_score = metrics.accuracy_score(y_test, cart_pred)
print(f"CART Accuracy: {cart_score:.4f} (~{cart_score*100:.2f}%)")
print("\nReport:")
print(metrics.classification_report(y_test, cart_pred))
```

```python
print("Phạm Lê Anh Khoa - 6451071035")

cart_cm = metrics.confusion_matrix(y_test, cart_pred)

plt.figure(figsize=(7, 6))
sns.heatmap(cart_cm, annot=True, fmt="d", square=True, cmap='Blues_r',
            xticklabels=class_names, yticklabels=class_names)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title(f'CART Decision Tree Accuracy Score: {cart_score:.4f}', size=13)
plt.show()
```

```python
print("Phạm Lê Anh Khoa - 6451071035")

fig, ax = plt.subplots(figsize=(24, 10))
tree.plot_tree(
    clf_cart, 
    max_depth=5, 
    feature_names=features_onehot.columns, 
    class_names=['e', 'p'], 
    filled=True, 
    fontsize=9
)
plt.savefig('mushroom_decision_tree_cart.png', dpi=100, bbox_inches='tight')
plt.show()
```

### Câu 9: Xây dựng mô hình phân lớp bằng thuật toán Naïve Bayes và kiểm tra kết quả đạt được

```python
print("Phạm Lê Anh Khoa - 6451071035")

gnb = GaussianNB()
gnb.fit(X_train, y_train)
bayes_pred = gnb.predict(X_test)

bayes_score = metrics.accuracy_score(y_test, bayes_pred)
print(f"Naive Bayes Accuracy: {bayes_score:.4f} (~{bayes_score*100:.2f}%)")
print("\nReport:")
print(metrics.classification_report(y_test, bayes_pred))
```

```python
print("Phạm Lê Anh Khoa - 6451071035")

bayes_cm = metrics.confusion_matrix(y_test, bayes_pred)

plt.figure(figsize=(7, 6))
sns.heatmap(bayes_cm, annot=True, fmt="d", square=True, cmap='Greens',
            xticklabels=class_names, yticklabels=class_names)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title(f'Naive Bayes Accuracy Score: {bayes_score:.4f}', size=13)
plt.show()
```

### Câu 10: So sánh kết quả của các mô hình trên

```python
print("Phạm Lê Anh Khoa - 6451071035")

print("=== BẢNG SO SÁNH ĐỘ CHÍNH XÁC (ACCURACY) TRÊN DỮ LIỆU MUSHROOM ===")
print(f"1. Cây quyết định ID3 (Entropy) : {id3_score:.4f} (~{id3_score*100:.2f}%)")
print(f"2. Cây quyết định CART (Gini)   : {cart_score:.4f} (~{cart_score*100:.2f}%)")
print(f"3. Naïve Bayes (GaussianNB)     : {bayes_score:.4f} (~{bayes_score*100:.2f}%)")
```

**Nhận xét và so sánh các mô hình:**
1. **Về độ chính xác tổng thể (Accuracy):**
   - Cả hai mô hình Cây quyết định **ID3** và **CART** đều đạt độ chính xác tuyệt đối **100% (1.0000)** trên tập kiểm thử.
   - Mô hình **Naïve Bayes (GaussianNB)** đạt độ chính xác cao **96.43%**, tuy nhiên vẫn có một số mẫu bị dự đoán nhầm do giả định tính độc lập có điều kiện giữa các thuộc tính chưa hoàn toàn thỏa mãn trên tập dữ liệu này.
2. **Đặc thù bài toán phân loại nấm độc:**
   - Cây quyết định rất hiệu quả với dữ liệu phân loại thuần túy (categorical), cho ra các luật rẽ nhánh rõ ràng (đặc biệt là thuộc tính mùi `odor` có khả năng phân biệt nấm độc rất cao).
   - Độ chính xác 100% của cây quyết định đảm bảo an toàn tuyệt đối khi phân loại nấm ăn được và nấm độc.
