**Họ tên:** Phạm Lê Anh Khoa  
**MSSV:** 6451071035  
**Dữ liệu:** Adult Income  

Code bám sát phần **Hướng dẫn trang 17–23** của tài liệu LAB2. Đặt `adult.data.csv` và `adult.test.csv` cùng thư mục notebook.

### Chuẩn bị – Import thư viện (trang 17)

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics
```

### Yêu cầu 1 – Nhập dữ liệu, đếm số dòng huấn luyện và kiểm thử

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

columns = ['age', 'workclass', 'fnlwgt', 'education', 'education-num',
           'marital-status', 'occupation', 'relationship', 'race', 'sex',
           'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income']

train = pd.read_csv('Dataset/adult.data.csv', names=columns, skipinitialspace=True, na_values='?')
test = pd.read_csv('Dataset/adult.test.csv', names=columns, skiprows=1, skipinitialspace=True, na_values='?')

print('Number of training data:', len(train))
print('Number of test data:', len(test))
```

### Yêu cầu 2 – Xóa dòng có `?`, bỏ `fnlwgt`, nối dữ liệu


```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

train = train.dropna()
test = test.dropna()

train = train.drop(columns=['fnlwgt'])
test = test.drop(columns=['fnlwgt'])

test['income'] = test['income'].str.rstrip('.')

data = pd.concat([train, test], ignore_index=True)

print("Train:", train.shape)
print("Test:", test.shape)
print("Data:", data.shape)
data.info()
```

### Yêu cầu 3 – Khảo sát độ tương đồng Pearson bằng heatmap (trang 17–18)

Giữ đúng hai dòng mẫu; thêm `numeric_only=True` để Pandas mới bỏ qua cột dạng chữ.

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

print("Train:", train.shape)
print("Test:", test.shape)
print("Data:", data.shape)
print("Các cột số:", data.select_dtypes(include='number').columns.tolist())
print("Ma trận Pearson:", data.corr(method='pearson', numeric_only=True).shape)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

plt.figure(figsize=(16,9))
sns.heatmap(data.corr(method='pearson', numeric_only=True), annot=True)
```

**Nhận xét theo tài liệu:** Các thuộc tính số không có độ tương quan cao, nên không cần loại bỏ cột nào.

### Yêu cầu 4 – Tách `features` và `labels` (trang 18)

```python
features = data.drop('income', axis=1)
labels = data['income']
```

### Yêu cầu 5 – One-hot encoding (trang 18–19)

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

print(features.select_dtypes(exclude=['int64']).columns)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

features_onehot = pd.get_dummies(features, columns=features.select_dtypes(exclude=['int64']).columns)
print(features_onehot)
```

### Yêu cầu 6 – Tách dữ liệu huấn luyện và kiểm thử (trang 19)

Sử dụng số dòng huấn luyện sau xử lý (`len(train)`), tương đương `30162` nếu đúng dữ liệu chuẩn trong tài liệu.

```python
X_train = features_onehot[:len(train)]
X_test = features_onehot[len(train):]
y_train = labels[:len(train)]
y_test = labels[len(train):]
```

### Yêu cầu 7 – Cây ID3 (`criterion="entropy"`) (trang 19–21)

```python
clf = tree.DecisionTreeClassifier(criterion="entropy", random_state=0)
clf.fit(X_train, y_train)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

# Predict the response for test dataset
tree_pred = clf.predict(X_test)
# Model Accuracy, how often is the classifier correct?
tree_score = metrics.accuracy_score(y_test, tree_pred)
print('Accuracy:', tree_score)
print('Report:', metrics.classification_report(y_test, tree_pred))
```

```python
tree_cm = metrics.confusion_matrix(y_test, tree_pred)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

plt.figure(figsize=(12,12))
sns.heatmap(tree_cm, annot=True, fmt='.3f', linewidths=.5, square=True, cmap='Blues_r')
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
title = 'Decision Tree Accuracy Score: {0}'.format(tree_score)
plt.title(title, size=15)
plt.show()
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

fig, ax = plt.subplots(figsize=(50, 24))
tree.plot_tree(clf, filled=True, fontsize=10)
plt.savefig('decision_tree.png', dpi=100)
plt.show()
```

### Yêu cầu 8 – Cây CART (`criterion="gini"`) (trang 22)

```python
clf_cart = tree.DecisionTreeClassifier(criterion="gini", random_state=0)
clf_cart.fit(X_train, y_train)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

cart_pred = clf_cart.predict(X_test)
cart_score = metrics.accuracy_score(y_test, cart_pred)
print('Accuracy:', cart_score)
print('Report:', metrics.classification_report(y_test, cart_pred))
```

```python
cart_cm = metrics.confusion_matrix(y_test, cart_pred)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

plt.figure(figsize=(12,12))
sns.heatmap(cart_cm, annot=True, fmt='.3f', linewidths=.5, square=True, cmap='Blues_r')
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title('CART Accuracy Score: {0}'.format(cart_score), size=15)
plt.show()
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

fig, ax = plt.subplots(figsize=(50, 24))
tree.plot_tree(clf_cart, filled=True, fontsize=10)
plt.savefig('decision_tree_cart.png', dpi=100)
plt.show()
```

### Yêu cầu 9 – Naïve Bayes (`GaussianNB`) (trang 22–23)

```python
gnb = GaussianNB()
bayes_pred = gnb.fit(X_train, y_train).predict(X_test)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

bayes_score = metrics.accuracy_score(y_test, bayes_pred)
print('Accuracy:', bayes_score)
print('Report:', metrics.classification_report(y_test, bayes_pred))
```

```python
bayes_cm = metrics.confusion_matrix(y_test, bayes_pred)
```

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

plt.figure(figsize=(12,12))
sns.heatmap(bayes_cm, annot=True, fmt='.3f', linewidths=.5, square=True, cmap='Greens')
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title('Naive Bayes Accuracy Score: {0}'.format(bayes_score), size=15)
plt.show()
```

### Yêu cầu 10 – So sánh kết quả ID3, CART và Naïve Bayes (trang 23)

```python
print("Họ tên: Phạm Lê Anh Khoa | MSSV: 6451071035")

print('Accuracy ID3:', tree_score)
print('Accuracy CART:', cart_score)
print('Accuracy Naive Bayes:', bayes_score)

ket_qua = {'ID3': tree_score, 'CART': cart_score, 'Naive Bayes': bayes_score}
print('Mô hình có Accuracy cao nhất:', max(ket_qua, key=ket_qua.get))
```

**Nhận xét:** So sánh Accuracy và các chỉ số precision, recall, f1-score từ ba báo cáo đã in ở trên. Mô hình có Accuracy cao nhất được xác định theo kết quả chạy thực tế, không dùng số liệu cố định trong ví dụ của tài liệu.
