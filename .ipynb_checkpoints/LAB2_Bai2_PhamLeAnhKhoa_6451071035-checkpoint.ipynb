## Chuẩn bị – Import thư viện (theo trang 17)

```python
# Nếu máy chưa cài thư viện, chạy một lần dòng sau (bỏ dấu #):
# %pip install numpy pandas matplotlib seaborn scikit-learn

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn import tree, metrics
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB

HO_TEN = "Phạm Lê Anh Khoa"
MSSV = "6451071035"

def in_thong_tin():
    print(f"Họ tên: {HO_TEN} | MSSV: {MSSV}")

in_thong_tin()
print("Đã import thành công các thư viện.")
```

## Yêu cầu 1 – Nhập dữ liệu và đếm số dòng ban đầu

```python
# Tên 15 cột chuẩn của bộ dữ liệu Adult (UCI).
columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race', 'sex',
    'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income'
]

# Chỉ cần đặt hai file này cạnh notebook, hoặc đổi đường dẫn dưới đây.
TRAIN_FILE = Path('Dataset/adult.data.csv')
TEST_FILE = Path('Dataset/adult.test.csv')


def doc_adult(path, url, is_test=False):
    """Đọc file địa phương (có/không header), hoặc file chuẩn UCI qua Internet."""
    source = path if path.exists() else url
    try:
        raw = pd.read_csv(source, header=None, skipinitialspace=True,
                          comment='|', na_values=['?', ' ?'], engine='python')
    except Exception as exc:
        raise FileNotFoundError(
            f"Không đọc được {path}. Hãy đặt file vào cùng thư mục notebook "
            f"hoặc mở kết nối Internet để tải từ UCI. Chi tiết: {exc}"
        ) from exc
    # File có dòng tiêu đề cột: loại bỏ header khỏi dữ liệu và lấy tên cột đó.
    first = [str(v).strip().lower() for v in raw.iloc[0].tolist()]
    if 'age' in first and ('income' in first or 'class' in first):
        raw.columns = [str(v).strip().lower() for v in raw.iloc[0].tolist()]
        raw = raw.iloc[1:].reset_index(drop=True)
    elif raw.shape[1] == len(columns):
        raw.columns = columns
    elif raw.shape[1] == len(columns) - 1:
        raw.columns = [c for c in columns if c != 'fnlwgt']
    else:
        raise ValueError(f"File {source} có {raw.shape[1]} cột (mong đợi 14 hoặc 15).")
    raw.columns = raw.columns.str.strip().str.lower()
    if 'class' in raw.columns and 'income' not in raw.columns:
        raw = raw.rename(columns={'class': 'income'})
    for col in raw.select_dtypes(include='object').columns:
        raw[col] = raw[col].astype(str).str.strip()
        raw[col] = raw[col].replace({'nan': np.nan, '?': np.nan})
    # Nhãn test UCI có dấu chấm cuối, ví dụ >50K.
    raw['income'] = raw['income'].str.rstrip('.')
    return raw.reset_index(drop=True)

adult_train_raw = doc_adult(TRAIN_FILE,
    'https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data')
adult_test_raw = doc_adult(TEST_FILE,
    'https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.test', is_test=True)

in_thong_tin()
print('Số lượng dữ liệu huấn luyện ban đầu:', len(adult_train_raw))
print('Số lượng dữ liệu kiểm thử ban đầu:', len(adult_test_raw))
print('Các cột dữ liệu huấn luyện:', adult_train_raw.columns.tolist())
print('Các cột dữ liệu kiểm thử:', adult_test_raw.columns.tolist())
```

## Yêu cầu 2 – Xóa dòng chứa `?`, bỏ `fnlwgt`, nối dữ liệu

```python
# Xóa dữ liệu thiếu và bỏ cột fnlwgt; bảo đảm hai phần có cùng cấu trúc.
adult_train = adult_train_raw.replace(r'^\s*\?\s*$', np.nan, regex=True).dropna().copy()
adult_test = adult_test_raw.replace(r'^\s*\?\s*$', np.nan, regex=True).dropna().copy()
adult_train = adult_train.drop(columns=['fnlwgt'], errors='ignore')
adult_test = adult_test.drop(columns=['fnlwgt'], errors='ignore')

# Đồng bộ thứ tự cột trước khi ghép; đảm bảo tập test không bị trộn vào tập train.
if set(adult_train.columns) != set(adult_test.columns):
    raise ValueError('Cột ở train và test không khớp sau khi tiền xử lý.')
adult_test = adult_test[adult_train.columns]
train_count = len(adult_train)
test_count = len(adult_test)
data = pd.concat([adult_train, adult_test], ignore_index=True)

# Ép về kiểu số cho những thuộc tính số trong UCI.
numeric_columns = ['age', 'education-num', 'capital-gain', 'capital-loss', 'hours-per-week']
for col in numeric_columns:
    data[col] = pd.to_numeric(data[col], errors='raise')

in_thong_tin()
print('Số lượng dữ liệu huấn luyện sau làm sạch:', train_count)
print('Số lượng dữ liệu kiểm thử sau làm sạch:', test_count)
print('Tổng số dòng sau nối:', len(data))
print('Số giá trị thiếu:', int(data.isna().sum().sum()))
print('Thông tin dữ liệu gộp:')
data.info()
print('5 dòng đầu:')
print(data.head().to_string(index=False))
```

## Yêu cầu 3 – Hệ số tương quan Pearson và heatmap (trang 17–18)

Pearson chỉ áp dụng trực tiếp cho các cột dạng số. Tài liệu nhận xét các cột số không có tương quan cao và không phải bỏ cột nào. Đặt ngưỡng minh họa `0.90`; loại bỏ chỉ khi thực tế có cặp vượt ngưỡng.

```python
# Tương tự ví dụ sns.heatmap(data.corr(method='pearson'), annot=True)
# Chỉ lấy các cột số để tương thích pandas phiên bản mới.
corr = data.select_dtypes(include=np.number).corr(method='pearson')
threshold = 0.90
upper = corr.abs().where(np.triu(np.ones(corr.shape), k=1).astype(bool))
remove_columns = [col for col in upper.columns if any(upper[col] > threshold)]

in_thong_tin()
print('Ma trận tương quan Pearson:')
print(corr.round(4).to_string())
print(f'Các cột có tương quan tuyệt đối > {threshold} cần loại bỏ:', remove_columns or 'Không có')
plt.figure(figsize=(16, 9))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='rocket', linewidths=.5)
plt.title('Tương quan Pearson giữa các thuộc tính số')
plt.tight_layout()
plt.show()
if remove_columns:
    data = data.drop(columns=remove_columns)
```

## Yêu cầu 4 – Tách thuộc tính thường `features` và nhãn `labels` (trang 18)

```python
features = data.drop('income', axis=1)
labels = data['income']

in_thong_tin()
print('Kích thước features:', features.shape)
print('Kích thước labels:', labels.shape)
print('Phân bố nhãn:')
print(labels.value_counts().to_string())
```

## Yêu cầu 5 – Chuyển các cột dạng chữ sang one-hot (trang 18–19)

```python
# Theo hướng dẫn: chọn cột không phải số rồi pd.get_dummies.
categorical_columns = features.select_dtypes(exclude=np.number).columns.tolist()
features_onehot = pd.get_dummies(features, columns=categorical_columns, dtype=int)

in_thong_tin()
print('Các thuộc tính được one-hot:', categorical_columns)
print('Kích thước dữ liệu sau one-hot:', features_onehot.shape)
print('5 dòng đầu của dữ liệu mã hóa:')
print(features_onehot.head().to_string(index=False, max_cols=18))
```

## Yêu cầu 6 – Tách train/test đúng ranh giới ban đầu (trang 19)

Do loại bỏ các dòng chứa `?`, phải dùng **số dòng sau làm sạch** thay vì cố định `30162`, `15060` để vẫn đúng nếu dữ liệu đầu vào thay đổi.

```python
X_train = features_onehot.iloc[:train_count]
X_test = features_onehot.iloc[train_count:train_count + test_count]
y_train = labels.iloc[:train_count]
y_test = labels.iloc[train_count:train_count + test_count]

assert len(X_train) == len(y_train) == train_count
assert len(X_test) == len(y_test) == test_count
assert not X_train.isna().any().any() and not X_test.isna().any().any()

in_thong_tin()
print('X_train:', X_train.shape, '| y_train:', y_train.shape)
print('X_test :', X_test.shape, '| y_test :', y_test.shape)
```

## Yêu cầu 7 – Cây ID3 (`criterion="entropy"`) và đánh giá (trang 19–21)

```python
clf = DecisionTreeClassifier(criterion='entropy', random_state=0)
clf.fit(X_train, y_train)
tree_pred = clf.predict(X_test)
tree_score = metrics.accuracy_score(y_test, tree_pred)
tree_cm = metrics.confusion_matrix(y_test, tree_pred, labels=['<=50K', '>50K'])

in_thong_tin()
print('ID3 (entropy) - Accuracy:', tree_score)
print('Classification report:')
print(metrics.classification_report(y_test, tree_pred, zero_division=0))
print('Confusion matrix:')
print(tree_cm)
plt.figure(figsize=(12, 9))
sns.heatmap(tree_cm, annot=True, fmt='d', linewidths=.5, square=True, cmap='Blues',
            xticklabels=['<=50K', '>50K'], yticklabels=['<=50K', '>50K'])
plt.xlabel('Predicted label')
plt.ylabel('Actual label')
plt.title(f'ID3 (entropy) - Accuracy: {tree_score:.4f}')
plt.tight_layout()
plt.show()
```

```python
# Biểu diễn cây theo tree.plot_tree() như tài liệu. Cây đầy đủ rất lớn,
# nên hiển thị 4 tầng đầu để xem rõ, đồng thời lưu hình cây đầy đủ riêng.
in_thong_tin()
print('ID3 - số nút:', clf.tree_.node_count, '| độ sâu:', clf.get_depth())
fig, ax = plt.subplots(figsize=(24, 10))
tree.plot_tree(clf, filled=True, fontsize=8, max_depth=3,
               feature_names=X_train.columns.tolist(), class_names=clf.classes_, ax=ax)
ax.set_title('Cây ID3 – 4 tầng đầu (cây đầy đủ được lưu dưới dạng PNG)')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(65, 28))
tree.plot_tree(clf, filled=True, fontsize=2,
               feature_names=X_train.columns.tolist(), class_names=clf.classes_, ax=ax)
fig.savefig('decision_tree_ID3_full.png', dpi=100, bbox_inches='tight')
plt.close(fig)
print('Đã lưu cây ID3 đầy đủ: decision_tree_ID3_full.png')
```

## Yêu cầu 8 – Cây CART (`criterion="gini"`) và đánh giá (trang 22)

```python
cart = DecisionTreeClassifier(criterion='gini', random_state=0)
cart.fit(X_train, y_train)
cart_pred = cart.predict(X_test)
cart_score = metrics.accuracy_score(y_test, cart_pred)
cart_cm = metrics.confusion_matrix(y_test, cart_pred, labels=['<=50K', '>50K'])

in_thong_tin()
print('CART (gini) - Accuracy:', cart_score)
print('Classification report:')
print(metrics.classification_report(y_test, cart_pred, zero_division=0))
print('Confusion matrix:')
print(cart_cm)
plt.figure(figsize=(12, 9))
sns.heatmap(cart_cm, annot=True, fmt='d', linewidths=.5, square=True, cmap='Oranges',
            xticklabels=['<=50K', '>50K'], yticklabels=['<=50K', '>50K'])
plt.xlabel('Predicted label')
plt.ylabel('Actual label')
plt.title(f'CART (gini) - Accuracy: {cart_score:.4f}')
plt.tight_layout()
plt.show()
```

```python
in_thong_tin()
print('CART - số nút:', cart.tree_.node_count, '| độ sâu:', cart.get_depth())
fig, ax = plt.subplots(figsize=(24, 10))
tree.plot_tree(cart, filled=True, fontsize=8, max_depth=3,
               feature_names=X_train.columns.tolist(), class_names=cart.classes_, ax=ax)
ax.set_title('Cây CART – 4 tầng đầu (cây đầy đủ được lưu dưới dạng PNG)')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(65, 28))
tree.plot_tree(cart, filled=True, fontsize=2,
               feature_names=X_train.columns.tolist(), class_names=cart.classes_, ax=ax)
fig.savefig('decision_tree_CART_full.png', dpi=100, bbox_inches='tight')
plt.close(fig)
print('Đã lưu cây CART đầy đủ: decision_tree_CART_full.png')
```

## Yêu cầu 9 – Naïve Bayes với `GaussianNB` và đánh giá (trang 22–23)

```python
gnb = GaussianNB()
bayes_pred = gnb.fit(X_train, y_train).predict(X_test)
bayes_score = metrics.accuracy_score(y_test, bayes_pred)
bayes_cm = metrics.confusion_matrix(y_test, bayes_pred, labels=['<=50K', '>50K'])

in_thong_tin()
print('Naïve Bayes - Accuracy:', bayes_score)
print('Classification report:')
print(metrics.classification_report(y_test, bayes_pred, zero_division=0))
print('Confusion matrix:')
print(bayes_cm)
plt.figure(figsize=(12, 9))
sns.heatmap(bayes_cm, annot=True, fmt='d', linewidths=.5, square=True, cmap='Greens',
            xticklabels=['<=50K', '>50K'], yticklabels=['<=50K', '>50K'])
plt.xlabel('Predicted label')
plt.ylabel('Actual label')
plt.title(f'Naive Bayes - Accuracy: {bayes_score:.4f}')
plt.tight_layout()
plt.show()
```

## Yêu cầu 10 – So sánh ID3, CART và Naïve Bayes (trang 23)

Bảng so sánh dựa trên **cùng một tập kiểm thử**. Các số liệu chỉ xuất hiện sau khi chạy thực tế, không chép cố định từ ảnh minh họa trong tài liệu.

```python
models_pred = {'ID3 (entropy)': tree_pred, 'CART (gini)': cart_pred,
               'Naïve Bayes (GaussianNB)': bayes_pred}
comparison = pd.DataFrame([
    {
        'Mô hình': name,
        'Accuracy': metrics.accuracy_score(y_test, pred),
        'Precision (weighted)': metrics.precision_score(y_test, pred, average='weighted', zero_division=0),
        'Recall (weighted)': metrics.recall_score(y_test, pred, average='weighted', zero_division=0),
        'F1-score (weighted)': metrics.f1_score(y_test, pred, average='weighted', zero_division=0),
    }
    for name, pred in models_pred.items()
]).sort_values('Accuracy', ascending=False).reset_index(drop=True)

in_thong_tin()
print('Bảng so sánh kết quả 3 mô hình:')
print(comparison.round(4).to_string(index=False))
print('\nMô hình có Accuracy cao nhất:', comparison.iloc[0]['Mô hình'])
if comparison['Accuracy'].max() == comparison['Accuracy'].min():
    print('Nhận xét: cả ba mô hình có Accuracy bằng nhau trên tập kiểm thử.')
else:
    print('Nhận xét: kết quả khác nhau trên cùng tập kiểm thử; cần cân nhắc cả Precision, Recall và F1-score.')

ax = comparison.set_index('Mô hình')[['Accuracy', 'Precision (weighted)',
                                     'Recall (weighted)', 'F1-score (weighted)']].plot.bar(
    figsize=(13, 6), rot=0, ylim=(0, 1.05), grid=True)
ax.set_title('So sánh kết quả phân lớp Adult Income')
ax.set_ylabel('Điểm đánh giá')
plt.tight_layout()
plt.show()
```
